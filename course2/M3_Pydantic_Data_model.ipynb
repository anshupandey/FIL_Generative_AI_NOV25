```python
from langchain_openai import AzureChatOpenAI
llm = AzureChatOpenAI(model='gpt-4o-mini')
```

```python
from langchain_core.prompts import ChatPromptTemplate
prompt = ChatPromptTemplate.from_template(""" for the provided user input, you need to analyze and provide
                                          1. product name
                                          2. review sentiment in one word: positive or negative
user input: {input}   """)

userinput = "I like the HP Envy x360 laptop, its good, has high performance. i storngly recommend the smae. "
output=llm.invoke(prompt.invoke({"input":userinput}))
output.content
```

```python
from pydantic import BaseModel,Field
class sentimentdatamodel(BaseModel):
    sentiment:str=Field(description="sentiment associated with input as: positive or negative")
    product_name:str=Field(description="name of product used in the user review only if specific product name is available otherwise NONE")
```

```python
from langchain_core.prompts import ChatPromptTemplate
prompt = ChatPromptTemplate.from_template(""" for the provided user input, you need to analyze and provide
                                          1. product name: only if specific product name (or brand name) is available otherwise NONE (note that product type is not name)
                                          2. review sentiment in one word: positive or negative
user input: {input}      """)

userinput = "I like the laptop, its good, has high performance. i storngly recommend the smae. "
st_llm = llm.with_structured_output(sentimentdatamodel)
output=st_llm.invoke(prompt.invoke({"input":userinput}))
```
